## 1. My lane (or freestyle) and why
I chose the Refresh / Content Opportunity Scoring lane because I want to investigate which high-visibility pages showing signs of decline should be reviewed first. This could help content strategists decide where to spend their limited review time. I am interested in this lane because thousands of pages cannot all be reviewed manually.


## 2. The question: decision, action, cost of a wrong call

If an important declining page is missed, it may not be reviewed in time and could continue losing search visibility and traffic.

## 3. Quick look at the data (2-3 real numbers)
The starter dataset contains 30,000 pages, and 16,262 of them show signs of decline. Even after limiting the results to declining pages with at least 100 impressions, 13,152 pages remain. This suggests that there are too many candidates to review manually, so a ranked list could help content strategists decide which pages to examine first.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Makayla-Kelly/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)

declining_pages = (
    df["trend_direction"] == "down"
).sum()

declining_with_visibility = (
    (df["trend_direction"] == "down")
    & (df["impressions_90d"] >= 100)
).sum()

print("Total pages:", total_pages)
print("Pages showing decline:", declining_pages)
print("Declining pages with at least 100 impressions:", declining_with_visibility)
```

## 4. Careful words: what I can and can't claim

My project will be able to identify observed patterns between page characteristics and declining performance and provide decision-support by ranking pages for human review. However, it will not be able to prove that updating a page causes its performance to improve or claim that the model can predict or explain Google’s algorithm.
